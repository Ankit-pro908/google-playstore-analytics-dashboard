```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.io as pio
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import nltk
import webbrowser
import os
```

```python
apps_df = pd.read_csv("Play Store Data.csv")
reviews_df = pd.read_csv("User Reviews.csv")
```

```python
apps_df.head()
```

```python
reviews_df.head()
```

```python
#Data Cleaning
apps_df = apps_df.dropna(subset=['Rating'])
for columns in apps_df.columns :
    apps_df[columns].fillna(apps_df[columns].mode()[0], inplace =True)
apps_df.drop_duplicates(inplace = True)
apps_df = apps_df[apps_df['Rating']<=5]
reviews_df.dropna(subset= ["Translated_Review"],inplace=True)
```

```python
apps_df["Installs"]=apps_df["Installs"].str.replace(',','').str.replace('+','').astype(int)

apps_df["Price"]=apps_df["Price"].str.replace('$','').astype(float)
```

```python
apps_df.dtypes
```

```python
merged_df = pd.merge(apps_df, reviews_df,on = 'App', how ='inner')
```

```python
merged_df.head()
```

```python
def convert_size(size):
    if 'M' in size:
        return float(size.replace('M',''))
    elif 'k' in size:
        return float(size.replace('k',''))/1024
    else:
        return np.nan
```

```python
apps_df['Size'] = apps_df['Size'].apply(convert_size)
```

```python
apps_df
```

```python
apps_df['Reviews'] = apps_df['Reviews'].astype(int)
```

```python
apps_df['Log_Installs'] = np.log(apps_df['Installs'])
apps_df['Log_Reviews'] = np.log(apps_df['Reviews'])
```

```python
apps_df
```

```python
apps_df.dtypes
```

```python
def rating_group(rating):
    if rating>=4:
        return 'Top Rated App'
    elif rating>=3:
        return 'Above Average'
    elif rating>=2:
        return 'Average'
    else:
        return 'Below Average'
apps_df['Rating_Group']=apps_df['Rating'].apply(rating_group)
```

```python
apps_df['Price'] = apps_df['Price'].astype(float)
```

```python
apps_df['Revenue'] = apps_df['Price'] * apps_df['Installs']
```

```python
plot_containers=""
```

```python
def save_plot_as_html(fig,filename,insight):
    global plot_containers
    filepath = os.path.join(html_files_path,filename)
    html_content=pio.to_html(fig, full_html=False,include_plotlyjs='inline')
    plot_containers += f"""
    <div class = "plot-container" id="{filename}" onclick = "openPlot('{filename}')">
    <div class="plot">{html_content}</div> 
    <div class = "insights">{insight}</div>
    </div>
    """
    fig.write_html(filepath, full_html=True,include_plotlyjs="inline")
```

```python
plot_width=400
plot_height=300
plot_bgcolor='black'
text_color='white'
title_font={'size':16}
axis_font={'size':12}
```

```python
category_counts = apps_df["Category"].value_counts().nlargest(10)
fig1=px.bar(
    x=category_counts.index,
    y=category_counts.values,
    labels={'x':'Category','y':'Count'},
    title='Top Categories on Play Store',
    color=category_counts.index,
    color_discrete_sequence=px.colors.sequential.Plasma,
    width=400,
    height=300
)
fig1.update_layout(
    plot_bgcolor='black',
    paper_bgcolor='black',
    font_color='white',
    title_font={'size':16},
    xaxis=dict(title_font={'size':12}),
    yaxis=dict(title_font={'size':12}),
    margin=dict(l=10,r=10,t=30,b=10)

    )
fig1.update_traces(marker=dict(line=dict(color="white",width=1)))
save_plot_as_html(fig1,"Category Graph 1.html","The top categories on playstore are dominated by tools, entertainment, and productivity apps")  
```
